```python
import numpy as np
```

```python
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
len_X_train = len(X_train)
len_X_test = len(X_test)
print(len_X_train,len_X_test)
```

```python
X_train = X_train / 255 # for better accurcy we need to divide with 255 by becomig range btw 0-1 which was easily recognized by system reason is below
X_test = X_test / 255
```

```python
X_train[0]
```

```python
plt.matshow(X_train[7654])
```

```python
y_train[7654]
```

```python
X_flattened_train = X_train.reshape(len_X_train,28*28)
X_flattened_test = X_test.reshape(len_X_test,28*28)
```

```python
X_flattened_train.shape
```

```python
X_flattened_train[0]
```

```python
from tensorflow import keras
model = keras.Sequential([
    keras.layers.Dense(10,input_shape=(784,),activation='sigmoid')
])
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy']) #Adam is the method that adjusts the weights of the neural network during training.#Loss = how wrong the model's prediction is.#"While training, also show me how many predictions are correct.
model.fit(X_flattened_train,y_train,epochs=5) #epochs=5 means the model will go through the entire training dataset 5 times.
```

```python
#here accuracy is not that high so we came up with the idea of scalling which converts the data into small ones range between 0-1 by dividing  andd set with 255
# X_train = X_train / 255
# X_test = X_test / 255
```

```python
model.evaluate(X_flattened_test, y_test)
```

```python
y_predicted = model.predict(X_flattened_test)
y_predicted[0]
```

```python
#it will give 10 different predictions for picture [0]
# np.argmax finds a maximum element from an array and returns the index of it

np.argmax(y_predicted[0])
```

```python
# whether you can its is right or not
plt.matshow(X_test[0])
```

```python
y_predicted_labels = [np.argmax(i) for i in y_predicted]
y_predicted_labels[:5]
```

```python
# Confusion matrix = a table that shows what the model got right and what it confused.
cm = tf.math.confusion_matrix(labels=y_test,predictions=y_predicted_labels)
cm
```

```python
# for more visualization we use matplotlib or seaborn
import seaborn as sn
import matplotlib.pyplot as plt

sn.heatmap(cm, annot=True, fmt='d')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
# for increase test accuracy we can add more dense layers

model = keras.Sequential([
    keras.layers.Dense(100, input_shape=(784,), activation='relu'),# rectified linear function  # If input is negative → output 0 ,,If input is positive → output stays the same
    keras.layers.Dense(10, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(X_flattened_train, y_train, epochs=5)
```
